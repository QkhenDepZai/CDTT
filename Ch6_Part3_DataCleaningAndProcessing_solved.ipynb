# Phần 1: Xử lý dữ liệu thiếu (Missing Data)

## Bài 1.1 - Phát hiện dữ liệu thiếu

1. Tạo một DataFrame với dữ liệu nhân viên công ty có các cột: `TenNV`, `Tuoi`, `Luong`, `PhongBan`, `KinhNghiem` với một số giá trị thiếu (None/NaN).

2. Sử dụng các phương thức `isna()`, `isnull()`, `notna()` để phát hiện dữ liệu thiếu.

3. Đếm số lượng dữ liệu thiếu theo từng cột và tổng số dữ liệu thiếu trong toàn bộ DataFrame.

4. Tìm các hàng có ít nhất 2 giá trị thiếu và hiển thị chúng.

```python
import pandas as pd
import numpy as np

# 1. Tạo DataFrame với dữ liệu thiếu
data_nhanvien = {
    'TenNV': ['Nguyễn Văn A', 'Trần Thị B', 'Lê Văn C', 'Phạm Thị D', 'Hoàng Văn E'],
    'Tuoi': [25, None, 30, 28, None],
    'Luong': [15000000, 18000000, None, 22000000, None],
    'PhongBan': ['IT', None, 'Marketing', 'IT', 'HR'],
    'KinhNghiem': [2, 5, None, 7, 1]
}

df_nhanvien = pd.DataFrame(data_nhanvien)
print("DataFrame nhân viên với dữ liệu thiếu:")
print(df_nhanvien)

# 2. Phát hiện dữ liệu thiếu
print("\n--- isna() ---")
print(df_nhanvien.isna())
print("\n--- isnull() (giống isna) ---")
print(df_nhanvien.isnull())
print("\n--- notna() (phủ định của isna) ---")
print(df_nhanvien.notna())

# 3. Đếm số lượng dữ liệu thiếu
print("\nSố giá trị thiếu theo từng cột:")
print(df_nhanvien.isna().sum())
print("\nTỷ lệ thiếu (%) theo cột:")
print((df_nhanvien.isna().mean() * 100).round(1))
print("\nTổng số giá trị thiếu toàn DataFrame:", df_nhanvien.isna().sum().sum())

# 4. Tìm các hàng có ít nhất 2 giá trị thiếu
so_thieu_moi_hang = df_nhanvien.isna().sum(axis=1)
print("\nSố giá trị thiếu theo từng hàng:")
print(so_thieu_moi_hang)
print("\nCác hàng có ít nhất 2 giá trị thiếu:")
print(df_nhanvien[so_thieu_moi_hang >= 2])
```

## Bài 1.2 - Xử lý dữ liệu thiếu bằng dropna

1. Sử dụng DataFrame từ bài 1.1, áp dụng phương thức `dropna()` với các tham số khác nhau:
   - Loại bỏ tất cả hàng có ít nhất 1 giá trị thiếu (`how='any'`)
   - Loại bỏ hàng chỉ khi tất cả giá trị đều thiếu (`how='all'`)
   - Loại bỏ cột có dữ liệu thiếu (`axis=1`)

2. Áp dụng `dropna()` chỉ trên một số cột cụ thể (`subset`).

3. So sánh kích thước DataFrame trước và sau khi áp dụng từng phương pháp.

```python
import pandas as pd
import numpy as np

#  Tạo DataFrame với dữ liệu thiếu
data_nhanvien = {
    'TenNV': ['Nguyễn Văn A', 'Trần Thị B', 'Lê Văn C', 'Phạm Thị D', 'Hoàng Văn E'],
    'Tuoi': [25, None, 30, 28, None],
    'Luong': [15000000, 18000000, None, 22000000, None],
    'PhongBan': ['IT', None, 'Marketing', 'IT', 'HR'],
    'KinhNghiem': [2, 5, None, 7, 1]
}

df_nhanvien = pd.DataFrame(data_nhanvien)

# Sử dụng DataFrame từ bài 1.1
print("DataFrame gốc:")
print(df_nhanvien)
print(f"Kích thước gốc: {df_nhanvien.shape}")

# 1. dropna() với các tham số khác nhau

# (a) Loại bỏ tất cả hàng có ít nhất 1 giá trị thiếu
df_any = df_nhanvien.dropna(how='any')
print("\n(a) dropna(how='any'):")
print(df_any)

# (b) Loại bỏ hàng chỉ khi tất cả giá trị đều thiếu (không có hàng nào như vậy -> giữ nguyên)
df_all = df_nhanvien.dropna(how='all')
print("\n(b) dropna(how='all'):")
print(df_all)

# (c) Loại bỏ cột có dữ liệu thiếu (cột TenNV là cột duy nhất không thiếu)
df_cot = df_nhanvien.dropna(axis=1)
print("\n(c) dropna(axis=1):")
print(df_cot)

# 2. dropna() chỉ trên một số cột cụ thể (ví dụ: Tuoi, Luong)
df_subset = df_nhanvien.dropna(subset=['Tuoi', 'Luong'])
print("\n2. dropna(subset=['Tuoi', 'Luong']):")
print(df_subset)

# 3. So sánh kích thước trước và sau
print("\n3. So sánh kích thước:")
print(f"Gốc                       : {df_nhanvien.shape}")
print(f"how='any'                 : {df_any.shape}")
print(f"how='all'                 : {df_all.shape}")
print(f"axis=1                    : {df_cot.shape}")
print(f"subset=['Tuoi', 'Luong']  : {df_subset.shape}")
```

## Bài 1.3 - Thay thế dữ liệu thiếu bằng fillna

1. Sử dụng DataFrame từ bài 1.1, thay thế dữ liệu thiếu bằng các phương pháp khác nhau:
   - Thay thế bằng giá trị cố định (0 cho số, "Không xác định" cho chuỗi)
   - Thay thế bằng giá trị trung bình (mean) cho các cột số
   - Thay thế bằng giá trị trung vị (median) cho các cột số
   - Thay thế bằng giá trị phổ biến nhất (mode) cho các cột phân loại

2. Tạo một DataFrame khác có dữ liệu chuỗi thời gian và áp dụng forward fill và backward fill.

3. So sánh kết quả của các phương pháp thay thế khác nhau.

```python
import pandas as pd
import numpy as np

# 1. Tạo DataFrame với dữ liệu thiếu
data_nhanvien = {
    'TenNV': ['Nguyễn Văn A', 'Trần Thị B', 'Lê Văn C', 'Phạm Thị D', 'Hoàng Văn E'],
    'Tuoi': [25, None, 30, 28, None],
    'Luong': [15000000, 18000000, None, 22000000, None],
    'PhongBan': ['IT', None, 'Marketing', 'IT', 'HR'],
    'KinhNghiem': [2, 5, None, 7, 1]
}

df_nhanvien = pd.DataFrame(data_nhanvien)

# 1. Thay thế bằng các phương pháp khác nhau
print("DataFrame gốc:")
print(df_nhanvien)

# Thay thế bằng giá trị cố định (0 cho số, "Không xác định" cho chuỗi)
df_codinh = df_nhanvien.fillna({'Tuoi': 0, 'Luong': 0, 'KinhNghiem': 0,
                                'PhongBan': 'Không xác định'})
print("\nfillna giá trị cố định:")
print(df_codinh)

# Thay thế bằng mean cho cột số
cot_so = ['Tuoi', 'Luong', 'KinhNghiem']
df_mean = df_nhanvien.copy()
df_mean[cot_so] = df_mean[cot_so].fillna(df_mean[cot_so].mean())
print("\nfillna bằng mean:")
print(df_mean)

# Thay thế bằng median cho cột số
df_median = df_nhanvien.copy()
df_median[cot_so] = df_median[cot_so].fillna(df_median[cot_so].median())
print("\nfillna bằng median:")
print(df_median)

# Thay thế bằng mode cho cột phân loại (mode() trả về Series -> lấy [0])
df_mode = df_nhanvien.copy()
df_mode['PhongBan'] = df_mode['PhongBan'].fillna(df_mode['PhongBan'].mode()[0])
print("\nfillna bằng mode (PhongBan):")
print(df_mode)

# 2. Tạo DataFrame chuỗi thời gian để demo forward/backward fill
data_timeseries = {
    'ngay': pd.date_range('2024-01-01', periods=7, freq='D'),
    'gia_co_phieu': [100, None, 105, None, None, 110, 115],
    'khoi_luong': [1000, 1200, None, 1500, None, None, 1800]
}

df_timeseries = pd.DataFrame(data_timeseries)
print("\nDataFrame chuỗi thời gian:")
print(df_timeseries)

# Forward fill và backward fill
df_ffill = df_timeseries.ffill()
df_bfill = df_timeseries.bfill()
print("\nForward fill (ffill):")
print(df_ffill)
print("\nBackward fill (bfill):")
print(df_bfill)

# 3. So sánh các phương pháp trên cột Luong và gia_co_phieu
so_sanh = pd.DataFrame({
    'Luong_goc': df_nhanvien['Luong'],
    'Luong_0': df_codinh['Luong'],
    'Luong_mean': df_mean['Luong'],
    'Luong_median': df_median['Luong'],
})
print("\nSo sánh các cách điền cho cột Luong:")
print(so_sanh)

so_sanh_ts = pd.DataFrame({
    'goc': df_timeseries['gia_co_phieu'],
    'ffill': df_ffill['gia_co_phieu'],
    'bfill': df_bfill['gia_co_phieu'],
    'noi_suy': df_timeseries['gia_co_phieu'].interpolate(),
})
print("\nSo sánh cho gia_co_phieu (kèm nội suy tuyến tính):")
print(so_sanh_ts)
```

# Phần 2: Xử lý dữ liệu trùng lặp (Duplicate Data)

## Bài 2.1 - Phát hiện và xử lý dữ liệu trùng lặp

1. Tạo một DataFrame về thông tin khách hàng có chứa dữ liệu trùng lặp với các cột: `TenKH`, `Tuoi`, `ThanhPho`, `SoDienThoai`.

2. Sử dụng phương thức `duplicated()` để phát hiện các hàng trùng lặp.

3. Đếm tổng số hàng trùng lặp và hiển thị các hàng bị trùng lặp.

4. Sử dụng `drop_duplicates()` để loại bỏ dữ liệu trùng lặp với các tham số khác nhau:
   - Giữ lại bản ghi đầu tiên (`keep='first'`)
   - Giữ lại bản ghi cuối cùng (`keep='last'`)
   - Loại bỏ tất cả bản ghi trùng lặp (`keep=False`)

5. Xử lý trùng lặp dựa trên một số cột cụ thể (`subset`).

```python
import pandas as pd

# 1. Tạo DataFrame có dữ liệu trùng lặp
data_khachhang = {
    'TenKH': ['Nguyễn Văn A', 'Trần Thị B', 'Nguyễn Văn A', 'Lê Văn C', 'Trần Thị B', 'Phạm Thị D'],
    'Tuoi': [25, 30, 25, 35, 30, 28],
    'ThanhPho': ['Hà Nội', 'TP.HCM', 'Hà Nội', 'Đà Nẵng', 'TP.HCM', 'Hà Nội'],
    'SoDienThoai': ['0123456789', '0987654321', '0123456789', '0111222333', '0987654321', '0999888777']
}

df_khachhang = pd.DataFrame(data_khachhang)
print("DataFrame khách hàng có dữ liệu trùng lặp:")
print(df_khachhang)

# 2. Phát hiện dữ liệu trùng lặp (mặc định keep='first': bản đầu không bị đánh dấu)
mask_trung = df_khachhang.duplicated()
print("\nduplicated():")
print(mask_trung)

# 3. Đếm và hiển thị các hàng trùng lặp
print("\nTổng số hàng trùng lặp:", mask_trung.sum())
print("Các hàng bị trùng (bản sao):")
print(df_khachhang[mask_trung])
print("Tất cả các hàng liên quan đến trùng lặp (keep=False):")
print(df_khachhang[df_khachhang.duplicated(keep=False)])

# 4. Loại bỏ dữ liệu trùng lặp với các tham số khác nhau
print("\nkeep='first':")
print(df_khachhang.drop_duplicates(keep='first'))
print("\nkeep='last':")
print(df_khachhang.drop_duplicates(keep='last'))
print("\nkeep=False (bỏ hết mọi bản ghi có trùng lặp):")
print(df_khachhang.drop_duplicates(keep=False))

# 5. Xử lý trùng lặp dựa trên subset
print("\nsubset=['SoDienThoai']:")
print(df_khachhang.drop_duplicates(subset=['SoDienThoai']))
print("\nsubset=['TenKH', 'ThanhPho'], keep='last', ignore_index=True:")
print(df_khachhang.drop_duplicates(subset=['TenKH', 'ThanhPho'], keep='last', ignore_index=True))
```

# Phần 3: Biến đổi và chuẩn hóa dữ liệu (Data Transformation)

## Bài 3.1 - Min-Max Normalization

1. Tạo một DataFrame về thông tin bất động sản với các cột: `DienTich` (m²), `Gia` (triệu VND), `SoPhong`, `TuoiNha` (năm).

2. Áp dụng Min-Max Normalization bằng công thức thủ công cho cột `Gia`.

3. Sử dụng `MinMaxScaler` từ scikit-learn để chuẩn hóa tất cả các cột số.

4. So sánh kết quả trước và sau khi chuẩn hóa bằng cách vẽ biểu đồ hoặc thống kê mô tả.

```python
from sklearn.preprocessing import MinMaxScaler
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Tạo DataFrame về bất động sản
data_batdongsan = {
    'DienTich': [50, 80, 120, 200, 300],
    'Gia': [2000, 3500, 5000, 8000, 12000],  # triệu VND
    'SoPhong': [2, 3, 4, 5, 6],
    'TuoiNha': [5, 10, 15, 20, 25]  # năm
}

df_batdongsan = pd.DataFrame(data_batdongsan)
print("DataFrame bất động sản gốc:")
print(df_batdongsan)
print("\nThống kê mô tả:")
print(df_batdongsan.describe())

# 2. Min-Max Normalization thủ công cho cột Gia: (x - min) / (max - min)
gia = df_batdongsan['Gia']
gia_minmax_tay = (gia - gia.min()) / (gia.max() - gia.min())
print("\nGia sau Min-Max thủ công:")
print(gia_minmax_tay)

# 3. Sử dụng MinMaxScaler cho tất cả cột số
scaler = MinMaxScaler()
arr = scaler.fit_transform(df_batdongsan)
df_minmax = pd.DataFrame(arr, columns=df_batdongsan.columns)
print("\nMinMaxScaler cho tất cả cột:")
print(df_minmax)
print("Khớp với công thức thủ công:", np.allclose(df_minmax['Gia'], gia_minmax_tay))

# 4. So sánh kết quả
print("\nThống kê sau chuẩn hóa:")
print(df_minmax.describe().loc[['min', 'max', 'mean', 'std']])

fig, axes = plt.subplots(1, 2, figsize=(11, 4))
df_batdongsan.plot(kind='box', ax=axes[0], title='Trước chuẩn hóa')
df_minmax.plot(kind='box', ax=axes[1], title='Sau Min-Max (đưa về [0, 1])')
plt.tight_layout()
plt.show()
```

## Bài 3.2 - Z-score Standardization

1. Sử dụng DataFrame từ bài 3.1, áp dụng Z-score Standardization bằng công thức thủ công cho cột `DienTich`.

2. Sử dụng `StandardScaler` từ scikit-learn để chuẩn hóa tất cả các cột số.

3. Kiểm tra xem dữ liệu sau standardization có mean ≈ 0 và std ≈ 1 không.

4. So sánh kết quả giữa Min-Max Normalization và Z-score Standardization.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler
import pandas as pd
import numpy as np

# 1. Tạo DataFrame về bất động sản
data_batdongsan = {
    'DienTich': [50, 80, 120, 200, 300],
    'Gia': [2000, 3500, 5000, 8000, 12000],  # triệu VND
    'SoPhong': [2, 3, 4, 5, 6],
    'TuoiNha': [5, 10, 15, 20, 25]  # năm
}

df_batdongsan = pd.DataFrame(data_batdongsan)
# 1. Z-score Standardization thủ công cho cột DienTich
print("DataFrame gốc:")
print(df_batdongsan)

# Tính Z-score thủ công cho DienTich: (x - mean) / std
# StandardScaler dùng độ lệch chuẩn tổng thể (ddof=0) nên phải truyền ddof=0 để khớp
dt = df_batdongsan['DienTich']
dt_z_tay = (dt - dt.mean()) / dt.std(ddof=0)
print("\nDienTich Z-score thủ công (ddof=0):")
print(dt_z_tay)

# 2. Sử dụng StandardScaler cho tất cả cột
std_scaler = StandardScaler()
df_std = pd.DataFrame(std_scaler.fit_transform(df_batdongsan), columns=df_batdongsan.columns)
print("\nStandardScaler cho tất cả cột:")
print(df_std)
print("Khớp công thức thủ công:", np.allclose(df_std['DienTich'], dt_z_tay))

# 3. Kiểm tra mean và std
print("\nMean sau standardization (≈ 0):")
print(df_std.mean().round(10))
print("Std sau standardization (≈ 1, ddof=0):")
print(df_std.std(ddof=0).round(10))
print("mean ≈ 0:", np.allclose(df_std.mean(), 0), "| std ≈ 1:", np.allclose(df_std.std(ddof=0), 1))

# 4. So sánh Min-Max vs Standardization
df_minmax = pd.DataFrame(MinMaxScaler().fit_transform(df_batdongsan), columns=df_batdongsan.columns)
print("\nMin-Max:")
print(df_minmax.round(3))
print("\nStandardization:")
print(df_std.round(3))
print("\nMiền giá trị Min-Max :", df_minmax.min().min(), "→", df_minmax.max().max())
print("Miền giá trị Z-score :", round(df_std.min().min(), 3), "→", round(df_std.max().max(), 3))
print("""
Nhận xét:
- Min-Max đưa dữ liệu về khoảng cố định [0, 1], giữ nguyên hình dạng phân phối, nhưng rất nhạy với outlier.
- Z-score đưa về mean = 0, std = 1, không bị chặn khoảng; phù hợp khi dữ liệu xấp xỉ phân phối chuẩn
  và với các thuật toán giả định dữ liệu tập trung quanh 0 (hồi quy, SVM, PCA, ...).
""")
```

## Bài 3.3 - Robust Scaler và xử lý Outliers

1. Tạo một DataFrame mới có chứa outliers (giá trị ngoại lai) trong cột `Gia`.

2. So sánh hiệu quả của 3 phương pháp scaling khi có outliers:
   - MinMaxScaler
   - StandardScaler  
   - RobustScaler

3. Sử dụng `RobustScaler` để xử lý dữ liệu có outliers.

4. Vẽ biểu đồ hoặc thống kê để thấy rõ sự khác biệt giữa các phương pháp.

```python
from sklearn.preprocessing import RobustScaler, MinMaxScaler, StandardScaler
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Tạo DataFrame có outliers
data_outliers = {
    'DienTich': [50, 80, 120, 200, 300, 95],
    'Gia': [2000, 3500, 5000, 8000, 50000, 4200],  # 50000 là outlier
    'SoPhong': [2, 3, 4, 5, 8, 3],  # 8 là outlier
    'TuoiNha': [5, 10, 15, 20, 25, 12]
}

df_outliers = pd.DataFrame(data_outliers)
print("DataFrame có outliers:")
print(df_outliers)
print("\nThống kê mô tả:")
print(df_outliers.describe())

# 2. So sánh 3 phương pháp scaling
cols = df_outliers.columns
df_minmax = pd.DataFrame(MinMaxScaler().fit_transform(df_outliers), columns=cols)
df_std = pd.DataFrame(StandardScaler().fit_transform(df_outliers), columns=cols)

# 3. Sử dụng RobustScaler: (x - median) / IQR
df_robust = pd.DataFrame(RobustScaler().fit_transform(df_outliers), columns=cols)

so_sanh_gia = pd.DataFrame({
    'Gia_goc': df_outliers['Gia'],
    'MinMax': df_minmax['Gia'],
    'Standard': df_std['Gia'],
    'Robust': df_robust['Gia'],
})
print("\nSo sánh 3 scaler trên cột Gia:")
print(so_sanh_gia.round(3))

# 4. Phân tích kết quả
print("\nTrong 5 giá trị bình thường của Gia, độ rộng sau scale:")
binh_thuong = so_sanh_gia.drop(index=4)
print((binh_thuong[['MinMax', 'Standard', 'Robust']].max() - binh_thuong[['MinMax', 'Standard', 'Robust']].min()).round(3))

fig, axes = plt.subplots(1, 4, figsize=(16, 4))
for ax, (ten, d) in zip(axes, [('Gốc', df_outliers), ('MinMax', df_minmax),
                               ('Standard', df_std), ('Robust', df_robust)]):
    d[['Gia']].plot(kind='box', ax=ax, title=ten)
plt.tight_layout()
plt.show()

print("""
Nhận xét:
- MinMax: outlier 50000 làm max rất lớn nên 5 giá trị còn lại bị nén vào khoảng [0, ~0.13].
- Standard: mean và std bị outlier kéo lệch nên các giá trị bình thường cũng bị nén (xấp xỉ -0.5).
- Robust: dùng median và IQR (ít bị ảnh hưởng bởi outlier) nên dữ liệu bình thường vẫn trải đều,
  còn outlier vẫn lộ rõ với giá trị rất lớn -> dễ phát hiện/xử lý riêng.
""")
```

# Phần 4: Xử lý chuỗi ký tự (String Processing)

## Bài 4.1 - Xử lý chuỗi ký tự cơ bản

1. Tạo một DataFrame về thông tin sản phẩm với các cột: `TenSP`, `MoTa`, `DanhMuc` có dữ liệu chuỗi không đồng nhất (có khoảng trắng thừa, chữ hoa/thường khác nhau).

   - `.str.strip()` để loại bỏ khoảng trắng
   - `.str.replace()` để thay thế ký tự
   - `.str.len()` để tính độ dài chuỗi

3. Tìm kiếm và lọc dữ liệu bằng:

```python
import pandas as pd

# 1. Tạo DataFrame với dữ liệu chuỗi không đồng nhất
data_sanpham = {
    'TenSP': ['  iPhone 15 Pro  ', 'samsung galaxy s23', 'XIAOMI Mi 13', '  MacBook Air M2  ', 'dell XPS 13'],
    'MoTa': ['Điện thoại thông minh cao cấp', '  Smartphone Android mới nhất  ', 'ĐIỆN THOẠI XIAOMI GIÁ RẺ', 'Laptop Apple siêu mỏng', '  máy tính xách tay Dell  '],
    'DanhMuc': ['Điện thoại/Apple', 'Điện thoại/Samsung', 'Điện thoại/Xiaomi', 'Laptop/Apple', 'Laptop/Dell']
}

df_sanpham = pd.DataFrame(data_sanpham)
print("DataFrame sản phẩm với dữ liệu chuỗi không đồng nhất:")
print(df_sanpham)

# 2. Xử lý chuỗi cơ bản
df_sach = df_sanpham.copy()
df_sach['TenSP_lower'] = df_sach['TenSP'].str.lower()
df_sach['TenSP_upper'] = df_sach['TenSP'].str.upper()
df_sach['TenSP_title'] = df_sach['TenSP'].str.strip().str.title()
print("\nlower / upper / title:")
print(df_sach[['TenSP', 'TenSP_lower', 'TenSP_upper', 'TenSP_title']])

# strip: loại bỏ khoảng trắng thừa ở hai đầu
df_sach['TenSP'] = df_sach['TenSP'].str.strip()
df_sach['MoTa'] = df_sach['MoTa'].str.strip()

# replace: thay thế ký tự/chuỗi
df_sach['DanhMuc_moi'] = df_sach['DanhMuc'].str.replace('/', ' - ', regex=False)
df_sach['MoTa_chuan'] = df_sach['MoTa'].str.lower().str.replace('điện thoại', 'ĐT', regex=False)

# len: độ dài chuỗi
df_sach['DoDai_TenSP'] = df_sach['TenSP'].str.len()
print("\nSau strip / replace / len:")
print(df_sach[['TenSP', 'DanhMuc_moi', 'MoTa_chuan', 'DoDai_TenSP']])

# 3. Tìm kiếm và lọc dữ liệu
print("\nSản phẩm có tên chứa 'phone' (không phân biệt hoa/thường):")
print(df_sach[df_sach['TenSP'].str.contains('phone', case=False)])

print("\nSản phẩm có mô tả chứa 'điện thoại' hoặc 'smartphone':")
print(df_sach[df_sach['MoTa'].str.contains('điện thoại|smartphone', case=False, regex=True)][['TenSP', 'MoTa']])

print("\nTên sản phẩm bắt đầu bằng 'mac' (không phân biệt hoa/thường):")
print(df_sach[df_sach['TenSP'].str.lower().str.startswith('mac')]['TenSP'])

print("\nTên sản phẩm kết thúc bằng '13':")
print(df_sach[df_sach['TenSP'].str.endswith('13')]['TenSP'])

# 4. Tách chuỗi
tach = df_sach['DanhMuc'].str.split('/', expand=True)
tach.columns = ['LoaiHang', 'ThuongHieu']
print("\nTách DanhMuc thành 2 cột:")
print(tach)
df_sach = pd.concat([df_sach, tach], axis=1)
print("\nSố sản phẩm theo loại hàng:")
print(df_sach['LoaiHang'].value_counts())
```

## Bài 4.2 - Regular Expressions (Regex)

1. Tạo một DataFrame chứa thông tin liên hệ khách hàng với cột `ThongTin` chứa địa chỉ email, số điện thoại, website.

2. Sử dụng regex để trích xuất:
   - Địa chỉ email bằng pattern cho email
   - Số điện thoại Việt Nam (các định dạng khác nhau)
   - Website/URL

3. Làm sạch và chuẩn hóa số điện thoại về định dạng thống nhất.

4. Tạo các cột riêng biệt cho email, số điện thoại, website đã được trích xuất.

```python
import re
import pandas as pd

# 1. Tạo DataFrame với thông tin liên hệ
data_lienhe = {
    'KhachHang': ['Công ty A', 'Khách hàng B', 'Doanh nghiệp C', 'Cá nhân D', 'Tổ chức E'],
    'ThongTin': [
        'Email: info@congtyA.com, SĐT: 0123-456-789, Website: https://congtyA.com',
        'Liên hệ: khachhangB@gmail.com hoặc gọi +84 987 654 321',
        'Hotline: (028) 3825-7863, web: www.doanhnghiepC.vn, mail: contact@doanhnghiepC.vn',
        'Phone: 0987.654.321, email: canhanD@yahoo.com',
        'Tel: 1900-1234, site: https://tochucE.org.vn, email: admin@tochucE.org.vn'
    ]
}

df_lienhe = pd.DataFrame(data_lienhe)
print("DataFrame thông tin liên hệ:")
print(df_lienhe)

# 2. Trích xuất thông tin bằng regex
pattern_email = r'[\w.+-]+@[\w-]+(?:\.[\w-]+)+'
# SĐT: +84..., 0xx... (có thể có dấu ngoặc), hoặc đầu số 1900xxxx; cho phép phân cách bằng space, '.', '-'
pattern_phone = r'(\+84[\d\s.\-]{8,}\d|\(?0\d[\d\s.\-()]{7,}\d|1900[\s.\-]?\d{4})'
pattern_url = r'(?:https?://|www\.)[^\s,]+'

df_lienhe['Email_tho'] = df_lienhe['ThongTin'].str.extract(f'({pattern_email})', expand=False)
df_lienhe['SDT_tho'] = df_lienhe['ThongTin'].str.extract(pattern_phone, expand=False)
df_lienhe['Website_tho'] = df_lienhe['ThongTin'].str.extract(f'({pattern_url})', expand=False)
print("\nTrích xuất thô:")
print(df_lienhe[['KhachHang', 'Email_tho', 'SDT_tho', 'Website_tho']])

# 3. Làm sạch và chuẩn hóa số điện thoại về định dạng thống nhất
def chuan_hoa_sdt(s):
    if pd.isna(s):
        return None
    so = re.sub(r'\D', '', s)          # chỉ giữ lại chữ số
    if so.startswith('84'):             # +84 987... -> 0987...
        so = '0' + so[2:]
    if len(so) == 10:                   # di động: 0987-654-321
        return f'{so[:4]}-{so[4:7]}-{so[7:]}'
    if len(so) == 11:                   # bàn (có mã vùng): (028) 3825-7863 -> 028-3825-7863
        return f'{so[:3]}-{so[3:7]}-{so[7:]}'
    if so.startswith('1900'):           # tổng đài: 1900-1234
        return f'{so[:4]}-{so[4:]}'
    return so

df_lienhe['SDT'] = df_lienhe['SDT_tho'].apply(chuan_hoa_sdt)

# 4. Tạo các cột riêng biệt đã chuẩn hóa
df_lienhe['Email'] = df_lienhe['Email_tho'].str.lower()
df_lienhe['Website'] = df_lienhe['Website_tho'].str.lower()
df_ket_qua = df_lienhe[['KhachHang', 'Email', 'SDT', 'Website']]
print("\nKết quả cuối cùng:")
print(df_ket_qua)

# Bonus: kiểm tra định dạng email bằng fullmatch
print("\nEmail hợp lệ?")
print(df_ket_qua['Email'].str.fullmatch(pattern_email))
```

# Phần 5: Xử lý dữ liệu phân loại (Categorical Data)

## Bài 5.1 - Label Encoding

1. Tạo một DataFrame về đánh giá khóa học với các cột: `TenKhoaHoc`, `TrinhDo` (Cơ bản, Trung cấp, Nâng cao), `XepLoai` (Kém, Trung bình, Khá, Giỏi, Xuất sắc), `NgonNgu`.

2. Áp dụng Label Encoding cho:
   - Cột `TrinhDo` (có thứ tự tự nhiên)
   - Cột `XepLoai` (có thứ tự tự nhiên)
   - Cột `NgonNgu` (không có thứ tự tự nhiên)

3. So sánh kết quả Label Encoding với pandas Category và thứ tự tùy chỉnh cho dữ liệu ordinal.

4. Phân tích ưu/nhược điểm của Label Encoding cho từng loại dữ liệu.

```python
from sklearn.preprocessing import LabelEncoder
import pandas as pd

# 1. Tạo DataFrame về đánh giá khóa học
data_khoahoc = {
    'TenKhoaHoc': ['Python cơ bản', 'Java nâng cao', 'SQL trung cấp', 'HTML cơ bản', 'React nâng cao'],
    'TrinhDo': ['Cơ bản', 'Nâng cao', 'Trung cấp', 'Cơ bản', 'Nâng cao'],
    'XepLoai': ['Khá', 'Xuất sắc', 'Giỏi', 'Trung bình', 'Giỏi'],
    'NgonNgu': ['Python', 'Java', 'SQL', 'HTML', 'JavaScript']
}

df_khoahoc = pd.DataFrame(data_khoahoc)
print("DataFrame đánh giá khóa học:")
print(df_khoahoc)

# 2. Áp dụng Label Encoding (sklearn gán nhãn theo thứ tự alphabet)
df_le = df_khoahoc.copy()
for cot in ['TrinhDo', 'XepLoai', 'NgonNgu']:
    le = LabelEncoder()
    df_le[cot + '_LE'] = le.fit_transform(df_le[cot])
    print(f"\n{cot}: {dict(zip(le.classes_, le.transform(le.classes_)))}")
print()
print(df_le)

# 3. Sử dụng pandas Category với thứ tự tùy chỉnh cho dữ liệu ordinal
thu_tu_trinh_do = ['Cơ bản', 'Trung cấp', 'Nâng cao']
thu_tu_xep_loai = ['Kém', 'Trung bình', 'Khá', 'Giỏi', 'Xuất sắc']

df_cat = df_khoahoc.copy()
df_cat['TrinhDo_cat'] = pd.Categorical(df_cat['TrinhDo'], categories=thu_tu_trinh_do, ordered=True)
df_cat['XepLoai_cat'] = pd.Categorical(df_cat['XepLoai'], categories=thu_tu_xep_loai, ordered=True)
df_cat['TrinhDo_code'] = df_cat['TrinhDo_cat'].cat.codes
df_cat['XepLoai_code'] = df_cat['XepLoai_cat'].cat.codes
print("\nPandas Category với thứ tự tùy chỉnh:")
print(df_cat[['TrinhDo', 'TrinhDo_code', 'XepLoai', 'XepLoai_code']])
print("\nCó thể so sánh vì ordered=True: XepLoai >= 'Giỏi'")
print(df_cat[df_cat['XepLoai_cat'] >= 'Giỏi'][['TenKhoaHoc', 'XepLoai']])

# So sánh LabelEncoder và Category tùy chỉnh
print("\nSo sánh TrinhDo:")
print(pd.DataFrame({'TrinhDo': df_khoahoc['TrinhDo'],
                    'LabelEncoder': df_le['TrinhDo_LE'],
                    'Category_tuy_chinh': df_cat['TrinhDo_code']}))

# 4. Phân tích kết quả
print("""
Phân tích:
- TrinhDo: LabelEncoder cho Cơ bản=0, Nâng cao=1, Trung cấp=2 (theo alphabet) -> SAI thứ tự thực
  (Trung cấp phải thấp hơn Nâng cao). Dùng Category/OrdinalEncoder với thứ tự tùy chỉnh: 0, 1, 2 đúng.
- XepLoai: LabelEncoder cho Giỏi=0, Khá=1, Trung bình=2, Xuất sắc=3 -> hoàn toàn lệch thứ tự. Cần thứ tự tùy chỉnh.
- NgonNgu: không có thứ tự; Label Encoding gán số 0..n-1 khiến mô hình hiểu nhầm có quan hệ lớn/nhỏ
  (SQL > Python?). Nên dùng One-Hot Encoding (Bài 5.2). Label Encoding chỉ phù hợp cho nhãn đích (target)
  hoặc mô hình dạng cây.
- Ưu điểm Label Encoding: đơn giản, gọn (1 cột), tiết kiệm bộ nhớ.
- Nhược điểm: tạo thứ tự giả cho dữ liệu nominal; sai thứ tự với ordinal nếu không chỉ định thủ công.
""")
```

## Bài 5.2 - One-Hot Encoding

1. Sử dụng DataFrame từ bài 5.1, áp dụng One-Hot Encoding cho các cột phân loại nominal (không có thứ tự tự nhiên).

2. So sánh hai phương pháp One-Hot Encoding:
   - Sử dụng `pd.get_dummies()`
   - Sử dụng `OneHotEncoder` từ scikit-learn

3. Xử lý tham số `drop_first=True` để tránh multicollinearity.

4. So sánh kích thước DataFrame trước và sau One-Hot Encoding và phân tích ưu/nhược điểm.

5. Kết hợp cả Label Encoding và One-Hot Encoding trong cùng một DataFrame.

```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd

# 1. Tạo DataFrame về đánh giá khóa học
data_khoahoc = {
    'TenKhoaHoc': ['Python cơ bản', 'Java nâng cao', 'SQL trung cấp', 'HTML cơ bản', 'React nâng cao'],
    'TrinhDo': ['Cơ bản', 'Nâng cao', 'Trung cấp', 'Cơ bản', 'Nâng cao'],
    'XepLoai': ['Khá', 'Xuất sắc', 'Giỏi', 'Trung bình', 'Giỏi'],
    'NgonNgu': ['Python', 'Java', 'SQL', 'HTML', 'JavaScript']
}

df_khoahoc = pd.DataFrame(data_khoahoc)
print("DataFrame đánh giá khóa học:")
print(df_khoahoc)

# Cột nominal (không có thứ tự): NgonNgu
# (TenKhoaHoc là định danh gần như duy nhất nên không mã hóa one-hot)
cot_nominal = ['NgonNgu']

# 2. One-Hot Encoding bằng pd.get_dummies()
df_dummies = pd.get_dummies(df_khoahoc, columns=cot_nominal, dtype=int)
print("\npd.get_dummies():")
print(df_dummies)

# 3. One-Hot Encoding bằng sklearn OneHotEncoder
ohe = OneHotEncoder(sparse_output=False)
arr = ohe.fit_transform(df_khoahoc[cot_nominal])
df_ohe = pd.DataFrame(arr, columns=ohe.get_feature_names_out(cot_nominal), index=df_khoahoc.index).astype(int)
print("\nOneHotEncoder (sklearn):")
print(df_ohe)
print("Hai cách cho cùng kết quả:", (df_ohe.values == pd.get_dummies(df_khoahoc[cot_nominal], dtype=int)[ohe.get_feature_names_out(cot_nominal)].values).all())

# Tham số drop_first=True để tránh multicollinearity (dummy variable trap)
df_dummies_drop = pd.get_dummies(df_khoahoc, columns=cot_nominal, drop_first=True, dtype=int)
print("\npd.get_dummies(drop_first=True):")
print(df_dummies_drop)

ohe_drop = OneHotEncoder(sparse_output=False, drop='first')
df_ohe_drop = pd.DataFrame(ohe_drop.fit_transform(df_khoahoc[cot_nominal]),
                           columns=ohe_drop.get_feature_names_out(cot_nominal)).astype(int)
print("\nOneHotEncoder(drop='first'):")
print(df_ohe_drop)

# 4. So sánh kích thước và phân tích
print("\nKích thước:")
print(f"Gốc                  : {df_khoahoc.shape}")
print(f"get_dummies          : {df_dummies.shape}")
print(f"get_dummies drop_first: {df_dummies_drop.shape}")
print("""
Phân tích:
- One-Hot không tạo thứ tự giả giữa các nhãn -> phù hợp dữ liệu nominal.
- Nhược điểm: số cột tăng theo số nhãn khác nhau (dữ liệu thưa, tốn bộ nhớ, dễ bị "curse of dimensionality").
- drop_first=True bỏ 1 cột (suy ra được từ các cột còn lại) để tránh đa cộng tuyến trong hồi quy tuyến tính.
- get_dummies tiện cho DataFrame; OneHotEncoder dùng được trong Pipeline và nhớ được các nhãn
  (handle_unknown='ignore' khi gặp nhãn mới ở dữ liệu test).
""")

# 5. Kết hợp Label Encoding (ordinal, thứ tự tùy chỉnh) và One-Hot Encoding (nominal)
df_ket_hop = df_khoahoc.copy()
df_ket_hop['TrinhDo'] = df_ket_hop['TrinhDo'].map({'Cơ bản': 0, 'Trung cấp': 1, 'Nâng cao': 2})
df_ket_hop['XepLoai'] = df_ket_hop['XepLoai'].map({'Kém': 0, 'Trung bình': 1, 'Khá': 2, 'Giỏi': 3, 'Xuất sắc': 4})
df_ket_hop = pd.get_dummies(df_ket_hop, columns=['NgonNgu'], dtype=int)
print("Kết hợp Label (ordinal) + One-Hot (nominal):")
print(df_ket_hop)
print("Kích thước:", df_ket_hop.shape)
```

# Phần 6: Bài tập tổng hợp

## Bài 6.1 - Dự án làm sạch dữ liệu hoàn chỉnh

**Mô tả:** Bạn được cung cấp dữ liệu khảo sát khách hàng của một cửa hàng online có nhiều vấn đề về chất lượng dữ liệu. Hãy áp dụng tất cả kiến thức đã học để làm sạch và chuẩn bị dữ liệu.

**Các bước thực hiện:**

1. **Tạo dữ liệu mẫu** có chứa tất cả các vấn đề:
   - Dữ liệu thiếu ở nhiều cột
   - Dữ liệu trùng lặp
   - Chuỗi ký tự không đồng nhất
   - Dữ liệu phân loại cần encoding
   - Dữ liệu số có outliers cần scaling

2. **Phân tích và báo cáo** tình trạng dữ liệu ban đầu

3. **Áp dụng các kỹ thuật làm sạch** đã học theo thứ tự phù hợp

4. **So sánh và đánh giá** kết quả trước/sau khi xử lý

5. **Xuất dữ liệu sạch** sẵn sàng cho phân tích/machine learning

```python
# Bước 1: Tạo dữ liệu mẫu có nhiều vấn đề
import re
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler, LabelEncoder, OneHotEncoder

# Tạo dữ liệu khảo sát khách hàng có vấn đề
data_khaosat = {
    'ho_ten': ['  Nguyễn Văn A  ', 'TRẦN THỊ B', 'nguyễn văn a', '  Lê Văn C  ', 'Phạm Thị D', 'TRẦN THỊ B', 'Hoàng Văn E', None],
    'tuoi': [25, None, 25, 35, None, 30, 28, 45],
    'gioi_tinh': ['Nam', 'Nữ', 'nam', 'Nam', 'Nữ', 'Nữ', None, 'Nam'],
    'thu_nhap': [15000000, 25000000, None, 35000000, 18000000, 25000000, 150000000, 22000000],  # 150M là outlier
    'muc_do_hai_long': ['Rất hài lòng', 'Hài lòng', None, 'Bình thường', 'Không hài lòng', 'Hài lòng', 'Rất hài lòng', 'Hài lòng'],
    'so_lan_mua_hang': [5, 3, 5, 12, None, 3, 20, 8],
    'email': ['  nguyena@gmail.com  ', 'tranthib@YAHOO.COM', None, 'levanc@outlook.com', '', 'tranthib@yahoo.com', 'hoange@company.vn', 'invalid-email'],
    'thanh_pho': ['Hà Nội', 'tp.hcm', 'Hà Nội', 'Đà Nẵng', 'TP.HCM', 'tp.hcm', None, 'Hải Phòng']
}

df_raw = pd.DataFrame(data_khaosat)
print("=== DỮ LIỆU KHẢO SÁT KHÁCH HÀNG (RAW) ===")
print(df_raw)

# Bước 2: Phân tích tình trạng dữ liệu ban đầu
print("\n=== PHÂN TÍCH DỮ LIỆU BAN ĐẦU ===")
print("Kích thước:", df_raw.shape)
print("\nKiểu dữ liệu:")
print(df_raw.dtypes)
print("\nSố giá trị thiếu theo cột:")
print(df_raw.isna().sum())
print("Email rỗng ('') :", (df_raw['email'] == '').sum())
print("\nSố hàng trùng lặp (duplicated() trên dữ liệu thô):", df_raw.duplicated().sum())
print("Giá trị gioi_tinh :", df_raw['gioi_tinh'].unique())
print("Giá trị thanh_pho :", df_raw['thanh_pho'].unique())

def so_outlier_iqr(s):
    q1, q3 = s.quantile([0.25, 0.75])
    iqr = q3 - q1
    return ((s < q1 - 1.5 * iqr) | (s > q3 + 1.5 * iqr)).sum()

print("\nSố outlier (IQR):")
for cot in ['tuoi', 'thu_nhap', 'so_lan_mua_hang']:
    print(f"  {cot}: {so_outlier_iqr(df_raw[cot])}")

# Bước 3: Áp dụng các kỹ thuật làm sạch
df = df_raw.copy()

# 3.1 Chuẩn hóa chuỗi (strip, thống nhất hoa/thường) - làm TRƯỚC khi tìm trùng lặp
df['ho_ten'] = df['ho_ten'].str.strip().str.title()
df['gioi_tinh'] = df['gioi_tinh'].str.strip().str.capitalize()
df['thanh_pho'] = df['thanh_pho'].str.strip().str.upper().replace({'TP.HCM': 'TP.HCM'})
df['thanh_pho'] = df['thanh_pho'].replace({'HÀ NỘI': 'Hà Nội', 'ĐÀ NẴNG': 'Đà Nẵng', 'HẢI PHÒNG': 'Hải Phòng'})
df['email'] = df['email'].str.strip().str.lower()

# 3.2 Kiểm tra email bằng regex: rỗng hoặc sai định dạng -> NaN
email_hop_le = df['email'].str.fullmatch(r'[\w.+-]+@[\w-]+(?:\.[\w-]+)+', na=False)
print("\nEmail không hợp lệ bị đặt về NaN:", df.loc[~email_hop_le & df['email'].notna(), 'email'].tolist())
df.loc[~email_hop_le, 'email'] = np.nan

# 3.3 Loại bỏ trùng lặp: cùng họ tên -> giữ bản ghi đầy đủ thông tin nhất
df['_so_thieu'] = df.isna().sum(axis=1)
df = (df.sort_values('_so_thieu', kind='stable')
        .drop_duplicates(subset=['ho_ten'], keep='first')
        .sort_index()
        .drop(columns='_so_thieu')
        .reset_index(drop=True))

# 3.4 Xử lý dữ liệu thiếu
for cot in ['tuoi', 'thu_nhap', 'so_lan_mua_hang']:      # số: median (bền vững với outlier)
    df[cot] = df[cot].fillna(df[cot].median())
df['gioi_tinh'] = df['gioi_tinh'].fillna(df['gioi_tinh'].mode()[0])
df['muc_do_hai_long'] = df['muc_do_hai_long'].fillna(df['muc_do_hai_long'].mode()[0])
df['thanh_pho'] = df['thanh_pho'].fillna('Không xác định')
df['email'] = df['email'].fillna('khong_co')
df['ho_ten'] = df['ho_ten'].fillna('Không xác định')

# 3.5 Mã hóa dữ liệu phân loại
df['muc_do_hai_long_code'] = df['muc_do_hai_long'].map(
    {'Không hài lòng': 0, 'Bình thường': 1, 'Hài lòng': 2, 'Rất hài lòng': 3})          # ordinal
df['gioi_tinh_code'] = LabelEncoder().fit_transform(df['gioi_tinh'])                    # nhị phân (Nam=0, Nữ=1)
df = pd.get_dummies(df, columns=['thanh_pho'], prefix='tp', dtype=int)                  # nominal -> one-hot

# 3.6 Scaling: outlier ở thu_nhap / so_lan_mua_hang -> RobustScaler; tuoi -> StandardScaler
df['thu_nhap_scaled'] = RobustScaler().fit_transform(df[['thu_nhap']])
df['so_lan_mua_hang_scaled'] = RobustScaler().fit_transform(df[['so_lan_mua_hang']])
df['tuoi_scaled'] = StandardScaler().fit_transform(df[['tuoi']])

# Bước 4: So sánh kết quả
print("\n=== SO SÁNH TRƯỚC / SAU ===")
print(f"Kích thước trước: {df_raw.shape} | sau: {df.shape}")
print("Tổng số giá trị thiếu trước:", df_raw.isna().sum().sum(), "| sau:", df.isna().sum().sum())
print("Số hàng trùng lặp (theo họ tên) sau:", df['ho_ten'].duplicated().sum())
print("\nDữ liệu sau khi làm sạch:")
print(df)
print("\nThống kê các cột đã scale:")
print(df[['tuoi_scaled', 'thu_nhap_scaled', 'so_lan_mua_hang_scaled']].describe().round(3))

# Bước 5: Xuất dữ liệu sạch
df.to_csv('khaosat_clean.csv', index=False, encoding='utf-8-sig')
print("\nĐã xuất dữ liệu sạch ra khaosat_clean.csv")
```
